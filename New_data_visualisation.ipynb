# List
## ordered collection of items stored in []
### mutable

```python
students=["Anish","Balaji","Chirag",'Devi','Neha']
print(students)
```

```python
# list with diffrent data types and access with their location
emp_info = [101, "Vansh","HR",85000.76]
print(emp_info)
print(emp_info[2])
print(emp_info[-3])
```

```python
# changing list of elements
cgpa=[7.2,6.4,8.9,9.5,7.4]
print(cgpa)
print("After updating")
cgpa[1]=9.9
print(cgpa)
```

```python
# adding one elemnet
students=[
    "Ganesh",
    'Hari',
    "Jhansi"
]
print(students)
students.append("Kiran")
print(students)
```

```python
# insert function in the list
students=["raj","Kumar","Chirag","Aneesh"]
students.insert(2,"Kiran")
print(students)
```

```python
# extend two list
section_a = ["Rahul","Sneha","Rishi"]
section_b=["Jhansi","Kishore","Nand","Meena"]
section_a.extend(section_b)
print(section_a)
print(section_b)
```

```python
# remove an item from list
students=["Rahul","Kiran","Sanjeev","Gowrish","Shivam"]
students.remove("Gowrish")
print(students)
```

```python
# remove an item by pop()
students=["Rahul","Kiran","Sanjeev","Gowrish","Yash"]
students.pop()
print(students)
```

```python
# list slicing
students=["Rahul","Kiran","Sanjeev","Gowrish","Yash"]
print(students[1:4])
# length of the list
print(len(students))
# arranging the data
students.sort()
print(students)
marks=[88,67,45,99,69]
marks.sort(reverse=True)
print("Decending_of_marks=",marks)
```

```python
# nested list
marks =[
    [91,88,87],
    [78,76,75],
    [67,87,58]
]
print(marks)
print(marks[0])
print(marks[1][2])
print(marks[2][1])
```
