```python
#create 2 table - Student Table
Student_info=pd.DataFrame({
    "Student_ID":["S101","S102","S103","S104"],
    "Name":["Anish","Riya","Manish kumar","Janvi"],
    "Department":["CSE","CSE","DS","DS"]
})
print(Student_info)
```

```python
# 2. Examination Table
exam = pd.DataFrame({
    "Student_ID":["S101","S102","S103","S105"],
    "Internal":[82,75,90,68],
    "Final":[88,72,94,70]
})
print(exam)
```

```python
#Merge - default - inner join
result=pd.merge(Student_info,exam,on="Student_ID",how="inner")
print(result)
```

```python
# Left join
Leftjoin_result=pd.merge(Student_info,exam,on="Student_ID",how="left")
print(Leftjoin_result)
```

```python
#Right_join
Rightjoin_result=pd.merge(Student_info,exam,on="Student_ID",how="right")
print(Rightjoin_result)
```

```python
# outerjoin
outerjoin_result=pd.merge(Student_info,exam,on="Student_ID",how="outer")
print(outerjoin_result)

```

```python
# Inner join - only command record (intersection)
# outer join - everything from both(union)
# left join - everything in the left + matchimg right
# right join - everything in the right + matching left
```

```python
# create 2 files
S_info=pd.DataFrame({
    "Student_ID":["S101","S102","S103","S104"],
    "Name":["Anish","Riya","Manish kumar","Janvi"],
    "Department":["CSE","CSE","DS","DS"]
})
exam1=pd.DataFrame({
    "ID":["S101","S102","S103"],
    "Marks":[88,72,95]
})

#merge 2 file - but no common column present in this
pd.merge(S_info,exam1,left_on="Student_ID",right_on="ID")
```
