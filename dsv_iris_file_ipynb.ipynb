```python
iris=pd.read_csv("/content/Iris.csv")
iris.head(3)
```

```python
print(iris.shape)
print(iris.columns.tolist())
print(iris.info())
```

```python
# file formats
iris.to_csv("iris_csvoutput.csv",index=False)

# load
iris_csv=pd.read_csv("/content/iris_csvoutput.csv")
iris_csv.head(7)
```

```python
iris.to_excel("iris_exceloutput.xlsx",index=False)
#load
iris_excel=pd.read_excel("/content/iris_exceloutput.xlsx")
iris_excel.tail(8)
```

```python
iris.to_json("iris_joutput.json",
             orient="records",
             indent=4)
# load and execute it
iris_json=pd.read_json("/content/iris_joutput.json")
iris_json.head(6)
```

```python
iris.to_parquet("iris_paroutput.parquet",index=False)
#loasd and display
iris_parquet=pd.read_parquet("/content/iris_paroutput.parquet")
iris_parquet.head(6)
```

```python
import os
files=[
    "iris_csvoutput.csv",
    "iris_exceloutput.xlsx",
    "iris_joutput.json",
    "iris_paroutput.parquet"
]
for file in files:
  size=os.path.getsize(file)
  print(f"{file}:{size} bytes")
```
