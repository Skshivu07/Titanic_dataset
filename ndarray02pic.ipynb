```python
# Reshaping the array
arr11=np.array([1,2,3,4,5,6])
new_arr=arr11.reshape(2,3)
print("Original array")
print(arr11)
print("\n Reshaped Array:")
print(new_arr)
```

```python
#Flattening
arr12=np.array([
    [1,2],
    [3,4],
    [5,6]
])
flat=arr12.flatten()
print("Original Array")
print(arr12)
print("\n Flattened Array:")
print(flat)
```

```python
# mathematical operations
marks=np.array([50,60,70,80,90])
print("Original Marks:",marks)
print("Adding + 5 Marks:",marks+5)
mul=marks*100
print("Multiplication with the array:",mul)
```

```python
#Aggregate Funtion
# sum
# min
#max
#avg
# count

arr31=np.array([45,67,78,89,56])
print(arr31)
print("Total:",arr31.sum())
print("\n Minimum value:",arr31.min())
print("\n Maximum value:",arr31.max())
print("\n Average value:",marks.mean())
```

```python
#Filtering
arr41=np.array([35,67,42,89,25,76])
result=arr41[arr41>=50]
print(result)
```

```python
#np.where
# to find positions satisfying a condition
marks=np.array([35,67,42,89,25,76])
positions=np.where(marks>=50)
print("Positions:",positions)
```

```python
#Sorting
sub_marks=np.array([35,67,42,89,25,76])
sorted_marks=np.sort(sub_marks)
print("Sorted Marks:",sorted_marks)
```

```python
#cumulative summation
patients=np.array([10,15,20,12])
cumulative_summation=patients.cumsum()
print("Daily Patients:",patients)
print("\n Patient for 4 days:",cumulative_summation)
```

```python
# for delete function
arr33=np.array([45,67,78,89,92])
del_arr33=np.delete(arr33,[1,3])
print(del_arr33)
```

```python
arr34=np.array([45,67,78,89,92])
reveresed_arr=np.flip(arr34)
print(reveresed_arr)
```

```python
#transpose matrix
arr35=np.array([
    [10,20,30],
    [40,50,60]
])
transposedarray=np.transpose(arr35)
print(transposedarray)
```
